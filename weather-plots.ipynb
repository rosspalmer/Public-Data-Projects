```python
import json
import plotly.graph_objects as go
import pandas as pd

sample = pd.read_parquet('/mnt/lake-fs/weather-sample')

id_denver = 117
id_dallas = 57
id_ashville = 86
id_houston = 820
id_charlotte = 825

cluster_id = id_dallas
rolling_n = 10
month = 5

json_string = sample[
    (sample['cluster_id'] == cluster_id) &
    (sample['rolling_n'] == rolling_n) &
    (sample['month'] == month)
]['json'].iloc[0]

data = json.loads(json_string)

# with open('./examples/hoston-july.json', 'r') as f:
#     r = f.read()
#     data = json.loads(r)
#     # del data["ghcn_id"]
#     # del data["month"]
    
#     df = pd.DataFrame(data)
#     df.set_index("year", inplace=True)
#     print(df.columns)
```

```python
layout = go.Layout(height=1000)

fig = go.Figure(layout=layout)

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_temperature'],
        name="average_daily_temperature",
        mode="markers",
        marker=go.scatter.Marker(color='green', opacity=0.2)
    )
)

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_temperature_avg'],
        name="average_daily_temperature_avg",
        marker=go.scatter.Marker(color='green')
    ))

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_min_temperature'],
        name="average_daily_min_temperature",
        mode="markers",
        marker=go.scatter.Marker(color='blue', opacity=0.2)
    )
)

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_min_temperature_avg'],
        name='average_daily_min_temperature_avg',
        marker=go.scatter.Marker(color='blue')
    ))

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_max_temperature'],
        name='average_daily_max_temperature',
        mode="markers",
        marker=go.scatter.Marker(color='red', opacity=0.2)
    )
)

fig.add_trace(
    go.Scatter(
        x=data['year'],
        y=data['average_daily_max_temperature_avg'],
        name='average_daily_max_temperature_avg',
        marker=go.scatter.Marker(color='red')
    ))

fig.show()
```
